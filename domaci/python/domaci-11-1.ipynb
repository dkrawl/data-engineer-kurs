```python
# 1.  Informacije o svemirskoj stanici Napisati funkciju
#     get_space_station() -\> dict koja preko Open Notify API-ja
#     (http://api.open-notify.org/iss-now.json) preuzima trenutnu poziciju
#     Međunarodne svemirske stanice (ISS) i vraća rečnik koji sadrži:
#     geografsku širinu (latitude), geografsku dužinu (longitude)

import requests

URL = 'http://api.open-notify.org/iss-now.json'

def get_space_station() -> dict:
    r = requests.get(URL)
    r.raise_for_status()
    return r.json()

get_space_station()
```

```python
# 2.  Pretraga univerziteta Napisati funkciju get_universities(country:
#     str) -\> list koja preko Hipolabs API-ja
#     (http://universities.hipolabs.com/search) preuzima univerzitete iz
#     zadate države i vraća listu njihovih naziva. Koristiti
#     params={‘country’: country} za prosleđivanje naziva države API-ju.

import requests

URL = 'http://universities.hipolabs.com/search'

def get_universities(country: str) -> list:
    r = requests.get(URL, params={'country': country})
    r.raise_for_status()
    return r.json()

get_universities('Sweden')
```

```python
# 3.  Informacije o kursu valuta Napisati funkciju get_exchange_rate(base:
#     str, target: str) -\> float koja preko Frankfurter API-ja
#     (https://api.frankfurter.dev/v1/latest) preuzima kurs između dve
#     valute. Funkcija treba da vrati koliko jedinica ciljne valute
#     (target) se dobija za jednu jedinicu početne valute (base).

import requests

URL = 'https://api.frankfurter.dev/v1/latest'

def get_exchange_rate(base: str, target: str) -> float:
    r = requests.get(URL)
    r.raise_for_status()

    exchange_rate = r.json()
    if base != 'EUR':
        return round(exchange_rate['rates'][target] / exchange_rate['rates'][base], 2)
    else:
        return exchange_rate['rates'][target]

get_exchange_rate('EUR', 'DKK')
```
