```python
# 1.  Sistem beleži broj aktivnih korisnika tokom 12 uzastopnih sati.
#     Potrebno je: Prikazati broj korisnika za peti sat. Prikazati broj
#     korisnika za poslednja četiri sata. Izračunati ukupan broj aktivnih
#     korisnika kroz svih 12 sati. Izračunati prosečan broj korisnika po
#     satu. Pronaći sat sa najvećim brojem korisnika.

import numpy as np
import math

active_users = np.array([
    20, 25, 31, 45, 100, 87, 45, 50, 66, 18, 23, 99
    ])

active_users_5th_hour = active_users[4]
active_users_last_4_hours = active_users[-4:].sum()
active_users_total = active_users.sum()
active_users_average = int(np.floor(np.mean(active_users)))
active_users_max_hour = np.argmax(active_users) + 1

print(active_users_5th_hour)
print(active_users_last_4_hours)
print(active_users_total)
print(active_users_average)
print(active_users_max_hour)
```

```python
# 2.  Sistem beleži CPU, RAM i disk usage za pet različitih servera.
#     Redovi predstavljaju servere, a kolone CPU, RAM i disk usage.
#     Potrebno je: Prikazati dimenzije matrice. Prikazati broj servera i
#     broj metrika. Izračunati prosečnu vrednost svake metrike. Izračunati
#     prosečnu potrošnju resursa za svaki server. Pronaći server sa
#     najvećom prosečnom potrošnjom resursa.

import numpy as np

server_usage = np.array([
    [45, 60, 70],
    [80, 75, 65],
    [30, 45, 50],
    [90, 85, 95],
    [55, 50, 40]
])

server_usage_dim = server_usage.shape
num_servers = server_usage.shape[0]
num_server_metrics = server_usage.shape[1]
server_metric_average = np.mean(server_usage, axis=0)
server_usage_average = np.mean(server_usage, axis=1)
server_max_usage = server_usage_average.argmax() + 1

print(server_usage_dim)
print(num_servers)
print(num_server_metrics)
print(server_metric_average)
print(server_usage_average)
print(server_max_usage)
```

```python
# 3.  Sistem beleži vreme izvršavanja 15 različitih procesa u sekundama.
#     Potrebno je: Pronaći sve procese koji su trajali duže od 100
#     sekundi. Pronaći procese koji su trajali između 40 i 100 sekundi.
#     Izbrojati procese koji su trajali kraće od 50 sekundi. Pronaći
#     indekse procesa koji su trajali duže od 100 sekundi. Izračunati
#     prosečno vreme izvršavanja svih procesa.

import numpy as np

process_time = np.array([
    65, 34, 84, 345, 234, 
    112, 543, 23, 68, 98, 
    99, 45, 323, 176, 5
    ])

process_time_greater_then_100 = process_time[process_time > 100]
process_time_between_40_and_100 = process_time[(process_time >= 40) & (process_time <= 100)]
num_processes_time_below_50 = process_time[np.where(process_time < 50)].size
indicies = np.where(process_time > 100)[0]
process_time_average = np.mean(process_time)

print(process_time_greater_then_100)
print(process_time_between_40_and_100)
print(num_processes_time_below_50)
print(indicies)
print(process_time_average)
```

```python
# 4.  Prodavnica beleži prihode od četiri različite kategorije proizvoda
#     tokom šest meseci. Redovi predstavljaju kategorije proizvoda, a
#     kolone mesece. Potrebno je: Izračunati ukupan prihod svake
#     kategorije proizvoda. Izračunati ukupan prihod za svaki mesec.
#     Pronaći kategoriju proizvoda sa najvećim ukupnim prihodom. Pronaći
#     mesec sa najvećim ukupnim prihodom. Izračunati prosečan prihod po
#     kategoriji proizvoda. Izračunati prosečan mesečni prihod svih
#     kategorija.

import numpy as np

revenue = np.array([
    [120000, 145000, 132000, 160000, 155000, 180000],
    [ 85000,  92000, 105000,  98000, 115000, 125000],
    [200000, 185000, 220000, 240000, 230000, 260000],
    [ 65000,  72000,  68000,  80000,  95000,  88000]
])

revenue_category = np.sum(revenue, axis=1)
revenue_month = np.sum(revenue, axis=0)
category_max_revenue = revenue_category.max()
month_max_revenue = revenue_month.max()
revenue_category_average = np.mean(revenue, axis=1)
revenue_category_average_per_month = np.mean(revenue, axis=0)

print(revenue_category)
print(revenue_month)
print(category_max_revenue)
print(month_max_revenue)
print(revenue_category_average)
print(revenue_category_average_per_month)
```

```python
# 5.  Sistem beleži prosečne dnevne temperature tokom 15 dana. Neki podaci
#     nedostaju ili nisu validni. Validna temperatura mora biti između -20
#     i 50 stepeni. Potrebno je: Identifikovati nedostajuće vrednosti.
#     Identifikovati nevalidne temperature (vrednosti manje od -20 ili
#     veće od 50). Pronaći indekse nevalidnih vrednosti. Izračunati broj
#     nedostajućih i nevalidnih vrednosti. Izračunati procenat nevalidnih
#     i nedostajućih zapisa zajedno. Kreirati novi niz koji sadrži samo
#     validne temperature. Izračunati prosečnu temperaturu bez
#     nedostajućih i nevalidnih vrednosti.

import numpy as np

daily_temp_avg = np.array([
    -15, -70, -22, -4, np.nan,
    24, 35, 8, 7, 15,
    65, 45, np.nan, 52, 12
    ])

missing_values = daily_temp_avg[np.isnan(daily_temp_avg)]
nonmissing_temp = daily_temp_avg[~np.isnan(daily_temp_avg)]
invalid_temp = nonmissing_temp[(nonmissing_temp < -20) | (nonmissing_temp > 50)]
invalid_temp_indicies = np.where((daily_temp_avg < -20) | (daily_temp_avg > 50))[0]
num_invalid_and_missing= invalid_temp.size + missing_values.size
pct_invalid_and_missing = num_invalid_and_missing / daily_temp_avg.size * 100
valid_temp = nonmissing_temp[(nonmissing_temp >= -20) & (nonmissing_temp <= 50)]
valid_temp_avg = round(np.mean(valid_temp), 2)

print(missing_values)
print(invalid_temp)
print(invalid_temp_indicies)
print(num_invalid_and_missing)
print(pct_invalid_and_missing)
print(valid_temp)
print(valid_temp_avg)
```
